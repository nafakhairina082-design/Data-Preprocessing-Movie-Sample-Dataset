**4. Transformasi Data**

a. Memastikan kolom (gross, budget, duration, imdb_score) sudah bertipe numerik

```python
# Pastikan kolom numerik dikonversi ke float
df['Gross'] = pd.to_numeric(df['Gross'], errors='coerce')
df['Budget'] = pd.to_numeric(df['Budget'], errors='coerce')
df['Duration'] = pd.to_numeric(df['Duration'], errors='coerce')
df['Imdb_Score'] = pd.to_numeric(df['Imdb_Score'], errors='coerce')

# Cek tipe data untuk memastikan
print(df.dtypes[['Gross', 'Budget', 'Duration', 'Imdb_Score']])
```

b. Memisahkan Kolom Genres Menjadi List

```python
if 'Genres' in df.columns:
    df['Genres'] = df['Genres'].astype(str)
    df['Genres_List'] = df['Genres'].str.split('|')

    # ambil semua genre unik
    all_genres = set([g for sublist in df['Genres_List'] for g in sublist if g != 'nan'])

    # buat kolom Genre_<nama>
    for genre in all_genres:
        col_name = "Genre_" + genre.strip()
        df[col_name] = df['Genres_List'].apply(lambda x: 1 if genre in x else 0)

    # hapus kolom Genres asli
    df = df.drop(columns=['Genres','Genres_List'])
df
```
